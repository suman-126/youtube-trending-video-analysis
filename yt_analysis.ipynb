```python
#Upload time analysis
#Convert the data column to datetime
df['publish_time']=pd.to_datetime(df['publish_time']) 
#Extract the hour
df['publish_hour']=df['publish_time'].dt.hour 
#Check the which time videos were published more
df['publish_hour'].value_counts().sort_index() 
```

```python
#Create the upload time heatmap
df['publish_day']=df['publish_time'].dt.day_name() #It extract the day of the week from publish time
#Checks how many videos were published on each day
df['publish_day'].value_counts()
```

```python
day_order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
df['publish_day'] = pd.Categorical(
    df['publish_day'],
    categories=day_order,
    ordered=True
)
#Create day + hour table
#Pd crosstab creates a table 
heatmap_data = pd.crosstab(  
    df['publish_day'],
    df['publish_hour']
) 
#WE write here filtered_df to change the heatmap according to the category.
```

```python
%pip install seaborn 
```

```python
#Display the heatmap
import seaborn as sns

plt.figure(figsize=(15,7))
sns.heatmap(heatmap_data)
plt.title('Youtube Video Upload Time Heatmap')
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.show()  
```
